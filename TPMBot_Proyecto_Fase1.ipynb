```python
# ============================================================
# PROYECTO FASE 1 - DESARROLLO DE APLICACIONES CON IA
# TPMBot - Asistente inteligente para Total Productive Maintenance
#
# Alumno: Apolinar Arguelles
#
# Arquitectura:
# Usuario -> Procesamiento NLP -> Identificacion de intencion
#         -> Base de conocimiento -> Respuesta
#
# Tecnologia:
# Python 3.10+
# NLTK
# Inteligencia Artificial simbolica basada en reglas
# ============================================================


# ------------------------------------------------------------
# 1. INSTALACION E IMPORTACION DE LIBRERIAS
# ------------------------------------------------------------

!pip -q install nltk

import nltk
import random
import re
import unicodedata

from nltk.chat.util import Chat, reflections


print("=" * 65)
print("TPMBot - Asistente de Total Productive Maintenance")
print("=" * 65)
print("NLTK cargado correctamente.")
print()


# ------------------------------------------------------------
# 2. FUNCIONES DE PROCESAMIENTO DE LENGUAJE
# ------------------------------------------------------------

def normalizar_texto(texto):
    """
    Normaliza la entrada del usuario.

    Convierte el texto a minusculas y elimina acentos
    para facilitar la comparacion mediante reglas.
    """

    texto = texto.lower().strip()

    texto = unicodedata.normalize("NFD", texto)

    texto = "".join(
        caracter
        for caracter in texto
        if unicodedata.category(caracter) != "Mn"
    )

    # Conservamos letras, numeros, espacios y algunos signos basicos.
    texto = re.sub(r"[^a-z0-9ñ\s\?\!\.\-]", "", texto)

    # Eliminar espacios repetidos.
    texto = re.sub(r"\s+", " ", texto)

    return texto


# ------------------------------------------------------------
# 3. BASE DE CONOCIMIENTO TPM
# ------------------------------------------------------------

conocimiento_tpm = {

    "tpm": [
        "TPM significa Total Productive Maintenance. "
        "Es una metodologia orientada a maximizar la efectividad de los "
        "equipos mediante la participacion de operadores, tecnicos, "
        "mantenimiento y otras areas de la organizacion.",

        "El objetivo de TPM es mejorar continuamente la disponibilidad, "
        "rendimiento y calidad de los procesos productivos, reduciendo "
        "fallas, defectos y perdidas."
    ],

    "mantenimiento_autonomo": [
        "El mantenimiento autonomo busca que el operador participe en "
        "actividades basicas como limpieza, inspeccion, lubricacion y "
        "deteccion temprana de anomalias.",

        "En mantenimiento autonomo, el operador ayuda a conservar las "
        "condiciones basicas del equipo y reporta las anomalias que "
        "requieren intervencion especializada."
    ],

    "mantenimiento_preventivo": [
        "El mantenimiento preventivo consiste en realizar actividades "
        "planificadas antes de que ocurra una falla, por ejemplo inspeccion, "
        "lubricacion, ajuste y reemplazo programado de componentes.",

        "La finalidad del mantenimiento preventivo es disminuir la "
        "probabilidad de falla mediante intervenciones planificadas."
    ],

    "mantenimiento_correctivo": [
        "El mantenimiento correctivo se realiza despues de que aparece "
        "una falla y tiene como objetivo restaurar la funcion del equipo.",

        "Cuando existe una falla que excede las actividades autorizadas "
        "para el operador, debe solicitarse apoyo de personal de mantenimiento."
    ],

    "oee": [
        "OEE significa Overall Equipment Effectiveness o Eficiencia Global "
        "del Equipo. Se analiza mediante tres factores: disponibilidad, "
        "rendimiento y calidad.",

        "La relacion general es: OEE = Disponibilidad x Rendimiento x Calidad."
    ],

    "5s": [
        "5S es una metodologia de organizacion del lugar de trabajo basada "
        "en clasificar, ordenar, limpiar, estandarizar y mantener la disciplina.",

        "Dentro de TPM, 5S ayuda a mantener el area organizada y facilita "
        "la identificacion visual de anomalias."
    ],

    "anomalia": [
        "Una anomalia puede ser una condicion diferente al funcionamiento "
        "normal del equipo, como ruido, vibracion, temperatura, fuga, "
        "desgaste u olor inusual.",

        "Cuando detectes una anomalia, evita intervenir fuera de tus "
        "responsabilidades. Registra la condicion y sigue el procedimiento "
        "establecido para escalarla a mantenimiento."
    ]

}


# ------------------------------------------------------------
# 4. DEFINICION DE INTENCIONES
# ------------------------------------------------------------
#
# Cada intencion contiene:
# - nombre
# - patrones o expresiones que puede utilizar el usuario
# - respuestas
#
# Esto representa el componente de IA simbolica del chatbot.
# ------------------------------------------------------------

intenciones = [

    # --------------------------------------------------------
    # SALUDO
    # --------------------------------------------------------

    {
        "intencion": "saludo",

        "patrones": [
            r"^(hola|buenos dias|buenas tardes|buenas noches|que tal|hey|saludos).*$"
        ],

        "respuestas": [
            "Hola. Soy TPMBot, tu asistente de Total Productive Maintenance. "
            "Puedo orientarte sobre TPM, OEE, mantenimiento autonomo, "
            "mantenimiento preventivo, 5S, inspecciones y anomalias.",

            "Hola. Bienvenido a TPMBot. ¿En que tema de mantenimiento o TPM puedo ayudarte?"
        ]
    },


    # --------------------------------------------------------
    # QUE ES TPM
    # --------------------------------------------------------

    {
        "intencion": "definicion_tpm",

        "patrones": [
            r".*(que es|que significa|explica|definicion de).*(tpm).*",
            r"^(tpm)$"
        ],

        "respuestas": conocimiento_tpm["tpm"]
    },


    # --------------------------------------------------------
    # OBJETIVO TPM
    # --------------------------------------------------------

    {
        "intencion": "objetivo_tpm",

        "patrones": [
            r".*(objetivo|finalidad|proposito).*(tpm).*",
            r".*(para que sirve).*(tpm).*"
        ],

        "respuestas": [
            "El objetivo de TPM es maximizar la efectividad de los equipos "
            "y reducir las perdidas asociadas con fallas, tiempos muertos, "
            "velocidad, defectos y otras condiciones que afectan la produccion."
        ]
    },


    # --------------------------------------------------------
    # MANTENIMIENTO AUTONOMO
    # --------------------------------------------------------

    {
        "intencion": "mantenimiento_autonomo",

        "patrones": [
            r".*(mantenimiento autonomo).*",
            r".*(actividades|tareas).*(operador).*",
            r".*(operador).*(limpieza|lubricacion|inspeccion).*"
        ],

        "respuestas": conocimiento_tpm["mantenimiento_autonomo"]
    },


    # --------------------------------------------------------
    # MANTENIMIENTO PREVENTIVO
    # --------------------------------------------------------

    {
        "intencion": "mantenimiento_preventivo",

        "patrones": [
            r".*(mantenimiento preventivo).*",
            r".*(prevenir|prevencion).*(falla|fallas).*",
            r".*(mantenimiento).*(programado|planificado).*"
        ],

        "respuestas": conocimiento_tpm["mantenimiento_preventivo"]
    },


    # --------------------------------------------------------
    # MANTENIMIENTO CORRECTIVO
    # --------------------------------------------------------

    {
        "intencion": "mantenimiento_correctivo",

        "patrones": [
            r".*(mantenimiento correctivo).*",
            r".*(reparar|reparacion).*(falla|maquina|equipo).*"
        ],

        "respuestas": conocimiento_tpm["mantenimiento_correctivo"]
    },


    # --------------------------------------------------------
    # OEE
    # --------------------------------------------------------

    {
        "intencion": "oee",

        "patrones": [
            r".*(que es|que significa|explica|calcular|calculo).*(oee).*",
            r".*(disponibilidad).*(rendimiento).*(calidad).*",
            r"^(oee)$"
        ],

        "respuestas": conocimiento_tpm["oee"]
    },


    # --------------------------------------------------------
    # 5S
    # --------------------------------------------------------

    {
        "intencion": "5s",

        "patrones": [
            r".*(que es|que son|explica|aplicar|metodologia).*(5s).*",
            r".*(clasificar|ordenar|limpiar).*(5s).*",
            r"^(5s)$"
        ],

        "respuestas": conocimiento_tpm["5s"]
    },


    # --------------------------------------------------------
    # VIBRACION
    # --------------------------------------------------------

    {
        "intencion": "anomalia_vibracion",

        "patrones": [
            r".*(vibracion|vibra|vibrando|vibraciones).*"
        ],

        "respuestas": [
            "Una vibracion anormal puede indicar una condicion que requiere "
            "evaluacion, por ejemplo desbalance, desalineacion, holgura o "
            "desgaste. No desmontes ni ajustes componentes si no estas "
            "autorizado. Registra la anomalia y sigue el procedimiento "
            "correspondiente para mantenimiento.",

            "Detectaste una posible anomalia de vibracion. Compara la condicion "
            "con el comportamiento normal del equipo, registra donde ocurre "
            "y escala la situacion si excede tus actividades autorizadas."
        ]
    },


    # --------------------------------------------------------
    # TEMPERATURA
    # --------------------------------------------------------

    {
        "intencion": "anomalia_temperatura",

        "patrones": [
            r".*(temperatura|caliente|calentamiento|sobrecalentamiento).*"
        ],

        "respuestas": [
            "Un incremento anormal de temperatura debe considerarse una "
            "anomalia. Evita manipular componentes calientes, registra la "
            "condicion observada y sigue el procedimiento de seguridad y "
            "escalamiento de mantenimiento aplicable al equipo.",

            "La temperatura fuera de la condicion normal puede relacionarse "
            "con friccion, lubricacion, carga, ventilacion u otras causas. "
            "TPMBot recomienda reportarla para diagnostico antes de realizar "
            "una intervencion tecnica."
        ]
    },


    # --------------------------------------------------------
    # RUIDO
    # --------------------------------------------------------

    {
        "intencion": "anomalia_ruido",

        "patrones": [
            r".*(ruido|ruidos|sonido raro|sonido anormal|suena raro).*"
        ],

        "respuestas": [
            "Un ruido diferente al funcionamiento normal puede ser una "
            "anomalia. Identifica en que zona se percibe, registra la "
            "condicion y reportala siguiendo el procedimiento de mantenimiento.",

            "Detectaste una posible anomalia de ruido. No intervengas "
            "mecanismos en movimiento. Documenta la ubicacion y las "
            "condiciones en las que aparece el ruido."
        ]
    },


    # --------------------------------------------------------
    # FUGAS
    # --------------------------------------------------------

    {
        "intencion": "anomalia_fuga",

        "patrones": [
            r".*(fuga|goteo|aceite tirado|derrame|escape de aire).*"
        ],

        "respuestas": [
            "Una fuga puede indicar perdida de aceite, aire, agua u otro "
            "fluido. Identifica visualmente la zona solo si puede hacerse "
            "de forma segura, evita contacto con sustancias desconocidas "
            "y reporta la anomalia.",

            "He identificado una posible fuga. Registra la ubicacion, "
            "el tipo de fluido si es conocido y la magnitud observable. "
            "Posteriormente sigue el procedimiento de escalamiento."
        ]
    },


    # --------------------------------------------------------
    # LUBRICACION
    # --------------------------------------------------------

    {
        "intencion": "lubricacion",

        "patrones": [
            r".*(lubricacion|lubricar|lubricante|grasa|engrasar).*"
        ],

        "respuestas": [
            "La lubricacion es una actividad importante dentro del "
            "mantenimiento. Debe utilizarse el lubricante, cantidad y "
            "frecuencia establecidos para cada equipo.",

            "Antes de lubricar un componente verifica que la actividad este "
            "autorizada, consulta el estandar del equipo y sigue las reglas "
            "de seguridad aplicables."
        ]
    },


    # --------------------------------------------------------
    # INSPECCION
    # --------------------------------------------------------

    {
        "intencion": "inspeccion",

        "patrones": [
            r".*(inspeccion|inspeccionar|revision|revisar equipo|checklist).*"
        ],

        "respuestas": [
            "Una inspeccion TPM puede incluir limpieza, condicion visual, "
            "fugas, ruidos, vibracion, temperatura, lubricacion, componentes "
            "flojos y condiciones basicas de seguridad.",

            "Para una inspeccion TPM comienza verificando las condiciones "
            "visuales y de seguridad. Despues revisa las condiciones "
            "definidas en el estandar o checklist autorizado del equipo."
        ]
    },


    # --------------------------------------------------------
    # PARO DE MAQUINA
    # --------------------------------------------------------

    {
        "intencion": "paro_maquina",

        "patrones": [
            r".*(maquina|equipo).*(paro|detenido|detuvo|no arranca|no funciona).*",
            r".*(paro de maquina|tiempo muerto).*"
        ],

        "respuestas": [
            "Si el equipo esta detenido, primero verifica las condiciones "
            "de seguridad y sigue el procedimiento autorizado para el equipo. "
            "Registra el paro y solicita soporte si la causa no corresponde "
            "a una actividad permitida para el operador.",

            "Un paro no planeado es una perdida importante dentro de TPM. "
            "Documenta la hora, equipo, sintoma observado y condicion del "
            "proceso para facilitar el diagnostico."
        ]
    },


    # --------------------------------------------------------
    # SEGURIDAD
    # --------------------------------------------------------

    {
        "intencion": "seguridad",

        "patrones": [
            r".*(seguridad|peligro|riesgo).*",
            r".*(bloqueo|etiquetado|loto).*"
        ],

        "respuestas": [
            "La seguridad tiene prioridad sobre produccion o mantenimiento. "
            "No realices intervenciones para las cuales no estes autorizado "
            "y sigue los procedimientos de seguridad establecidos en tu planta.",

            "Si existe una condicion insegura, no intentes resolverla fuera "
            "de tus responsabilidades. Sigue el procedimiento de seguridad "
            "y escalamiento establecido por la organizacion."
        ]
    },


    # --------------------------------------------------------
    # REPORTE DE ANOMALIA
    # --------------------------------------------------------

    {
        "intencion": "reporte_anomalia",

        "patrones": [
            r".*(reportar|reporte|registrar).*(anomalia|falla|problema).*",
            r".*(encontre|detecte|observe).*(anomalia|problema|falla).*"
        ],

        "respuestas": [
            "Para documentar una anomalia registra al menos: equipo, "
            "ubicacion, fecha u hora, sintoma observado y condicion en la "
            "que aparecio. Si tu proceso lo permite, agrega evidencia visual.",

            "Una buena descripcion de anomalia responde: ¿que equipo?, "
            "¿donde esta el problema?, ¿que observaste?, ¿cuando ocurrio? "
            "y ¿bajo que condicion estaba operando?"
        ]
    },


    # --------------------------------------------------------
    # AYUDA
    # --------------------------------------------------------

    {
        "intencion": "ayuda",

        "patrones": [
            r"^(ayuda|help|menu|opciones).*$",
            r".*(que puedes hacer|en que puedes ayudar).*"
        ],

        "respuestas": [
            """
Puedo ayudarte con los siguientes temas:

1. ¿Que es TPM?
2. Mantenimiento autonomo
3. Mantenimiento preventivo
4. Mantenimiento correctivo
5. OEE
6. Metodologia 5S
7. Inspecciones
8. Lubricacion
9. Vibraciones
10. Temperatura
11. Ruidos anormales
12. Fugas
13. Paros de maquina
14. Reporte de anomalias
15. Seguridad

Ejemplo:
"Mi maquina tiene una vibracion anormal"
            """
        ]
    },


    # --------------------------------------------------------
    # DESPEDIDA
    # --------------------------------------------------------

    {
        "intencion": "despedida",

        "patrones": [
            r"^(adios|bye|hasta luego|nos vemos|salir|fin|terminar).*$"
        ],

        "respuestas": [
            "Hasta luego. Recuerda: en TPM, detectar una anomalia a tiempo "
            "puede evitar una falla futura.",

            "Gracias por utilizar TPMBot. La mejora continua comienza "
            "con observar, detectar y actuar de manera estandarizada."
        ]
    }

]


# ------------------------------------------------------------
# 5. CONVERSION DE INTENCIONES AL FORMATO DE NLTK
# ------------------------------------------------------------

pares_nltk = []

for item in intenciones:

    for patron in item["patrones"]:

        # Agregamos identificador de intencion a la respuesta.
        respuestas_nltk = []

        for respuesta in item["respuestas"]:
            respuestas_nltk.append(
                f"[{item['intencion'].upper()}]\n{respuesta}"
            )

        pares_nltk.append(
            [
                patron,
                respuestas_nltk
            ]
        )


# Creamos el motor conversacional NLTK.
chat_nltk = Chat(pares_nltk, reflections)


# ------------------------------------------------------------
# 6. MOTOR PRINCIPAL DE TPMBot
# ------------------------------------------------------------

def responder(mensaje):

    texto = normalizar_texto(mensaje)

    if not texto:
        return (
            "ENTRADA NO VALIDA",
            "Por favor escribe una pregunta relacionada con TPM."
        )

    respuesta = chat_nltk.respond(texto)

    # --------------------------------------------------------
    # FALLBACK
    # --------------------------------------------------------

    if respuesta is None:

        return (
            "NO_RECONOCIDA",
            "No pude identificar con suficiente claridad tu consulta. "
            "Soy un chatbot especializado en TPM. Puedes preguntarme sobre "
            "mantenimiento autonomo, preventivo, correctivo, OEE, 5S, "
            "inspecciones, vibracion, temperatura, ruido, fugas, "
            "lubricacion o paros de maquina. Escribe 'ayuda' para ver "
            "las opciones."
        )

    # Recuperar identificador de intencion.
    coincidencia = re.match(
        r"\[([A-Z0-9_]+)\]\n(.*)",
        respuesta,
        re.DOTALL
    )

    if coincidencia:

        intencion = coincidencia.group(1).lower()
        texto_respuesta = coincidencia.group(2)

        return intencion, texto_respuesta

    return "respuesta_general", respuesta


# ------------------------------------------------------------
# 7. INTERFAZ DEL CHATBOT
# ------------------------------------------------------------

def iniciar_chatbot():

    print()
    print("=" * 65)
    print("                     TPMBot")
    print("      Asistente de Total Productive Maintenance")
    print("=" * 65)

    print("""
Hola. Soy TPMBot.

Soy un prototipo de chatbot basado en reglas desarrollado
con Python y NLTK.

Puedo orientarte sobre:
- TPM
- Mantenimiento autonomo
- Mantenimiento preventivo
- OEE
- 5S
- Inspecciones
- Anomalias
- Vibraciones
- Temperatura
- Fugas
- Ruidos
- Paros de maquina

Escribe "ayuda" para consultar el menu.
Escribe "salir" para terminar.
""")

    while True:

        try:

            mensaje = input("Usuario: ")

            texto_normalizado = normalizar_texto(mensaje)

            if texto_normalizado in [
                "salir",
                "fin",
                "terminar",
                "adios",
                "bye"
            ]:

                intencion, respuesta = responder(mensaje)

                print()
                print("TPMBot:", respuesta)
                print()
                break

            intencion, respuesta = responder(mensaje)

            print()
            print("TPMBot:", respuesta)

            # Mostrar intencion identificada.
            # Esto ayuda a demostrar el funcionamiento durante
            # las pruebas academicas.
            print(f"\n[Intencion identificada: {intencion}]")
            print()

        except KeyboardInterrupt:

            print("\n\nTPMBot: Conversacion finalizada.")
            break


# ------------------------------------------------------------
# 8. PRUEBAS AUTOMATICAS
# ------------------------------------------------------------

def ejecutar_pruebas():

    pruebas = [

        "Hola",

        "¿Que es TPM?",

        "¿Cual es el objetivo de TPM?",

        "Explica el mantenimiento autonomo",

        "¿Que es mantenimiento preventivo?",

        "¿Que significa OEE?",

        "¿Que son las 5S?",

        "Mi maquina tiene mucha vibracion",

        "El motor esta muy caliente",

        "La maquina tiene un ruido anormal",

        "Detecte una fuga de aceite",

        "¿Que debo revisar durante una inspeccion?",

        "La maquina se detuvo",

        "¿Como puedo reportar una anomalia?",

        "¿Que puedes hacer?",

        # Caso fuera del dominio para probar la limitacion.
        "¿Quien gano el partido de futbol?"
    ]

    print()
    print("=" * 70)
    print("PRUEBAS AUTOMATICAS DE TPMBot")
    print("=" * 70)

    exitos = 0
    no_reconocidas = 0

    for numero, pregunta in enumerate(pruebas, start=1):

        intencion, respuesta = responder(pregunta)

        print()
        print("-" * 70)

        print(f"PRUEBA {numero}")

        print()
        print("Entrada:")
        print(pregunta)

        print()
        print("Intencion identificada:")
        print(intencion)

        print()
        print("Respuesta:")
        print(respuesta)

        if intencion != "no_reconocida":
            exitos += 1
        else:
            no_reconocidas += 1

    print()
    print("=" * 70)
    print("RESUMEN")
    print("=" * 70)

    print(f"Total de conversaciones probadas: {len(pruebas)}")
    print(f"Entradas reconocidas: {exitos}")
    print(f"Entradas fuera del conocimiento: {no_reconocidas}")

    print()
    print("Las entradas no reconocidas demuestran una limitacion")
    print("esperada de un chatbot basado en reglas.")


# ------------------------------------------------------------
# 9. EJECUCION
# ------------------------------------------------------------

print()
print("TPMBot esta listo.")
print()
print("Funciones disponibles:")
print("  iniciar_chatbot()   -> inicia conversacion interactiva")
print("  ejecutar_pruebas()  -> ejecuta casos de prueba")
print()

# Inicia automaticamente el chatbot.
iniciar_chatbot()
```
